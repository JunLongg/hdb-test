## Data Quality Requirement 4
### Compute remaining lease rounded down to Years and Months.
#### Assumption: As the lease commencment date does not include the month of commencement, it is assumed to be 1st jan of the year, this is to ensure a more conservative narration than giving longer remaining lease 

```python
df
```

```python
# df['month'].astype('datetime64[D]')
# pd.to_datetime(df['lease_commence_date'], format='%Y')
df['computed_remaining_lease'] = 99 - ((df['month'].astype('datetime64[D]') - pd.to_datetime(df['lease_commence_date'], format='%Y')) / np.timedelta64(1,'Y'))
df['Years'] = np.floor(df['computed_remaining_lease']).astype('int')
df['Months'] = np.floor(np.modf(df['computed_remaining_lease'])[0] * 12).astype('int')
df['computed_remaining_lease_in_years_months'] = df['Years'].astype('str') + ' Years ' + df['Months'].astype('str') + ' Months'
df
```

## Data Quality Requirement 5
### Deduplication using all columns except resale price, taking the higher resale price when encountered.

```python
#creating list of columns to be ported over from previous worked on dataframe
# Making a new copy of the prepared dataframe without the validations
column_list = list(hdb_mar2012_dec2014.columns)
column_list.append('computed_remaining_lease_in_years_months')
df2 = df[column_list].copy()
df2
```

```python
dedupe_columns = [i for i in column_list if i != 'resale_price']
df2[df2.duplicated(subset=dedupe_columns)]
```

```python
# By using sort_values, the dataframe will be rearranged with the highest price above. Accompanied by the drop_duplicate function with its default keep = first configuration, this will ensure the deduplication will keep the highest resale value.
df2_deduped = df2.sort_values('resale_price', ascending=False).drop_duplicates(subset=dedupe_columns,ignore_index=True)
df2_deduped
```

```python
# Using z-score
from scipy import stats 
zscore = np.abs(stats.zscore(df2_deduped['resale_price']))
print('Z-Score values: \n', zscore)
```

```python
outliers_z = df2_deduped[zscore > 3]
outliers_z
```

```python
outliers_z.describe()
```

```python
# Using Interquartile Range
q1 = df2_deduped['resale_price'].quantile(0.25)
q3 = df2_deduped['resale_price'].quantile(0.75)
print(q1, q3)
iqr = q3 -q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers_iqr = df2_deduped[
        (df2_deduped['resale_price'] < lower_bound) | 
        (df2_deduped['resale_price'] > upper_bound)
    ]
print('Outliers using IQR: \n')
outliers_iqr
```

```python
df2_deduped.skew()
```

From above, using z-score, 1039 records are identified to be beyond 3 standard deviations from the mean value, ranging from 838,800 to 1,150,000 in terms of value, while 2842 records are identified using Interquartile range. As the resale_price column is skewed, z-score will not perform as well as IQR.

## Data Transformation Requirement
### create new Resale Identifier column

```python
df3 = df2_deduped.copy()
df3
```

```python
# First character: S.
# Next 3 digits: first 3 digits from block column, remove non-numerical characters, append zero if requrired
# Next 2 digits: 1st and 2nd digit of average price based on year-month, town, flat type grouping
# Last 2 digits: month of current entry
# Last char: first character of town
import re
df_price = df3.groupby(by=['computed_remaining_lease_in_years_months', 'town','flat_type'])[['resale_price']].mean().reset_index()
# df_price
def prepare_identifier(row):
    identifier = ''
    prep_blk = '0' * (3 - len(re.sub('[a-zA-Z]','',row['block']))) + re.sub('[a-zA-Z]','',row['block'])
    prep_price = str(df_price[
        (df_price['computed_remaining_lease_in_years_months'] == row['computed_remaining_lease_in_years_months'])  & 
        (df_price['town'] == row['town'])  & 
        (df_price['flat_type'] == row['flat_type'])].loc[:,'resale_price'])[:2]
    prep_month = row['month'][-2:]
    prep_town = row['town'][0]
    identifier = 'S' +prep_blk +prep_price + prep_month + prep_town
    return(identifier)
df3['resale_identifier'] = df3.apply(prepare_identifier, axis=1)
df3
```

```python
# Output dataframe with identifer to csv
df3.to_csv('.\Transformed\hdb_2012_2016_transformed.csv', index=False)
```

### Hashing

```python
df4 = df3.copy()
df4
```

```python
df4['resale_identifier'].nunique()
```

```python
# decided to use blake2b for its speed and modern security level
from hashlib import blake2b
def hashing(row):
    key = b'Password'
    h = blake2b(key=key)
    value = row['resale_identifier'].encode()
    h.update(value)
    return h.hexdigest()
df4['hashed_resale_identifier'] = df4.apply(hashing, axis=1)
df4
```

```python
# Output dataframe with hashed identifier to csv
df4.to_csv('.\Hashed\hdb_2012_2016_hashed.csv', index=False)
```
